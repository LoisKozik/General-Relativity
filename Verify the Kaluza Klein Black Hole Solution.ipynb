# Extremal Kaluza-Klein Black Hole Solution

```python
q = smp.Symbol('q', real=True, positive=True)
H = 1 + q/r
chi_expr = -smp.sqrt(3)/2 * smp.log(H) 
exp_fac = smp.exp(-smp.sqrt(3)*chi_expr)
```

```python
# Since the metric is not defined in the module we define it here.

K = smp.Matrix([
    [1/smp.sqrt(H), 0, 0, 0],
    [0, -smp.sqrt(H), 0, 0],               
    [0, 0, -r**2*smp.sqrt(H), 0],
    [0, 0, 0, -r**2*smp.sqrt(H)*(smp.sin(theta))**2]])
```

### Christoffel Symbols

```python
Gamma = Christoffel_Symbols(K)
Display_Christoffel_Symbols(K,Gamma)
```

### Riemann Tensor

```python
R = Riemann_Tensor(K,Gamma)
Display_Riemann_Tensor(K,R)
```

### Ricci Tensor

```python
Ric = Ricci_Tensor(K,R)
Display_Ricci_Tensor(K,R)
```

### Ricci Scalar

```python
Scal = Ricci_Scalar(K,R)
Display_Ricci_Scalar(K,R)
```
